# Modelo Regresión Lineal O3

#### con datos de ONI desde 2012

```r
# establecer fechas

year1 <- 2012
year2 <- as.numeric(format(Sys.Date(), "%Y")) -1
meses <- c(
  "enero", "febrero", "marzo", "abril", "mayo", "junio",
  "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"
)


# Clasificación de meses para ONI
oni_meses <- c(
  "DJF" = "enero",
  "JFM" = "febrero",
  "FMA" = "marzo",
  "MAM" = "abril",
  "AMJ" = "mayo",
  "MJJ" = "junio",
  "JJA" = "julio",
  "JAS" = "agosto",
  "ASO" = "septiembre",
  "SON" = "octubre",
  "OND" = "noviembre",
  "NDJ" = "diciembre"
)
```

```r
# Cargar bases

url_mei <- "https://psl.noaa.gov/enso/mei/data/meiv2.data"
nombres_bimestrales <- c(
  "DJ", "JF", "FM", "MA", "AM", "MJ", "JJ", "JA", "AS", "SO", "ON", "ND")
nombres_trimestrales <- c("DJF", "JFM", "FMA", "MAM", "AMJ", "MJJ", 
                          "JJA", "JAS", "ASO", "SON", "OND", "NDJ")


# Bases de ONI
oni <- read.table(
  "https://www.cpc.ncep.noaa.gov/data/indices/oni.ascii.txt",
  header = TRUE
) %>% 
  rename(Year = YR, meses = SEAS , oni = ANOM) %>% 
  filter(Year >= year1, Year < year(Sys.Date())) %>% 
  dplyr::select(- TOTAL) %>% 
  dplyr::mutate(oni = round(oni, 1),
         month = oni_meses[meses],
         month = factor(month, levels = .env$meses))


## Observados de ozono
o3.d <- evola.exp::redes_con(start = year1, end = year2) %>% 
  evola.exp::filtrar_est(transporte = F) %>% 
  dplyr::group_by(date = as.Date(date)) %>% 
  dplyr::summarise(valor = max(valor, na.rm = T))
```

```r
## Conteo de dias >= 155 por año
o3.m <- o3.d %>%
  mutate(
    Year = year(date),
    month_num = month(date),
    month = factor(meses[month_num], levels = meses)
  ) %>%
  group_by(Year, month_num, month) %>%
  summarise(
    ndias = sum(valor >= 155, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  arrange(Year, month_num) %>%
  dplyr::select(Year, month, ndias)
  # %>% filter(Year != 2024)
```

```r
# datos para modelos de regresión

# Se unen los dias de O3 >= 155 con datos de ONI
oni.conts <- o3.m %>%
  mutate(
    Year = as.numeric(Year),
    month = factor(month, levels = .env$meses)
  ) %>%
  left_join(
    oni %>% dplyr::select(Year, month, oni),
    by = c("Year", "month")
  )
```

### Datos para modelo de regresión:
##### Conteo mensual de días con O3 >= 155 ppb
##### Datos ONI

```r
# Revision de estructura de la base
head(oni.conts)
tail(oni.conts)
```

### Modelo de regresion lineal

```r


conts.lm <- lm(ndias ~ Year + month + oni, data = oni.conts)
```

```r
# Ver significancio del modelo lineal, residuos y ajuste del modelo
cat("===== Resumen del modelo =====\n")
summary(conts.lm)
cat("===== AIC =====\n")
AIC(conts.lm)
cat("===== BIC =====\n")
BIC(conts.lm)
```

```r
evalu <- oni.conts %>%
  mutate(
    pred = predict(
      conts.lm,
      newdata = oni.conts,
      type = "response"
    ),
    pred_round = round(pred),
    error = ndias - pred,
    abs_error = abs(error),
    error_cuad = error^2
  )

# Evaluar

MAE <- mean(evalu$abs_error, na.rm = TRUE)
RMSE <- sqrt(mean(evalu$error_cuad, na.rm = TRUE))

cat("\n MAE \n")
MAE
cat("\n RMSE \n")
RMSE
```

### Graficas de evaluacion

```r
par(mfrow = c(2, 2))
plot(conts.lm)
par(mfrow = c(1, 1))

cat("\n ===== Residuals vs Fitted ===== \n")
cat("\n No hay linealidad \n")

cat("\n ===== Q-Q Residuals ===== \n")
cat("\n Los residuos no parecen una distribución lineal \n")

cat("\n ===== Scale-Location ===== \n")
cat("\n Hay heterocedasticidad (varianza errores) \n")

cat("\n ===== Residuals vs Leverage ===== \n")
cat("\n Hay puntos influyentes que estan jalando el modelo (149, 17, 53) \n")
```

```r
# Breusch-Pagan Test (heterocedasticidad)
# H0: Hay homocedasticidad
# H1: Hay heterocedasticidad

cat("\n===== Breusch-Pagan Test (heterocedasticidad) =====\n")

bp_test <- bptest(conts.lm)

print(bp_test)

cat("\n===== Interpretación =====\n")

if (bp_test$p.value > 0.05) {
  cat("p-value =", bp_test$p.value, "\n")
  cat("Hay homocedasticidad, no se rechaza H0.\n")
} else {
  cat("p-value =", bp_test$p.value, "\n")
  cat("Hay heterocedasticidad, se rechaza H0.\n")
}
```

```r
cat("\n===== Verificar errores estandar robustos =====\n")

coeftest(conts.lm, vcov = vcovHC(conts.lm, type = "HC1"))
```

```r
# Durbin-Watson Test
# H0: No hay autocorrelación
# H1: Hay autocorrelación

cat("\n===== Durbin-Watson Test (autocorrelación) =====\n")

dw_test <- dwtest(conts.lm, alternative = "two.sided")

print(dw_test)

dw_value <- as.numeric(dw_test$statistic)
p_value <- dw_test$p.value

cat("\n===== Interpretación =====\n")
cat("DW =", round(dw_value, 4), "\n")
cat("p-value =", round(p_value, 4), "\n\n")

if (p_value > 0.05) {
  cat("No hay evidencia estadística de autocorrelación en los residuos. No se rechaza H0.\n")
} else {
  if (dw_value < 2) {
    cat("Hay evidencia de autocorrelación positiva en los residuos. Se rechaza H0.\n")
  } else if (dw_value > 2) {
    cat("Hay evidencia de autocorrelación negativa en los residuos. Se rechaza H0.\n")
  } else {
    cat("DW está muy cerca de 2, pero el p-value indica posible autocorrelación. Revisa el modelo.\n")
  }
}
```

##### El modelo no parece tener un problema grave de independencia temporal según Durbin–Watson, 
##### pero sí presenta heterocedasticidad fuerte según Breusch–Pagan. Por lo tanto, 
##### los coeficientes pueden seguir siendo útiles, pero la inferencia estadística del modelo 
##### debe interpretarse con cuidado o corregirse usando errores robustos.

```r
year1 <- 2012
year2 <- as.numeric(format(Sys.Date(), "%Y")) -1
meses <- c(
  "enero", "febrero", "marzo", "abril", "mayo", "junio",
  "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"
)


# Clasificación de meses para ONI
oni_meses <- c(
  "DJF" = "enero",
  "JFM" = "febrero",
  "FMA" = "marzo",
  "MAM" = "abril",
  "AMJ" = "mayo",
  "MJJ" = "junio",
  "JJA" = "julio",
  "JAS" = "agosto",
  "ASO" = "septiembre",
  "SON" = "octubre",
  "OND" = "noviembre",
  "NDJ" = "diciembre"
)

# Serie de tiempo
oni.ts <- ts(oni.conts$oni, start = c(year1, 1), end = c(year2, 12), frequency = 12)

# Modelo ARIMA para la serie de tiempo de ONI
# Se usa ARIMA para pronosticar los valores futuros de ONI
oni.ts.mod <- auto.arima(oni.ts, seasonal = T)

# Pronostico de ONI
oni.pron <- forecast(oni.ts.mod, h = 6)
```

```r
meses_pred <- c("febrero", "marzo", "abril", "mayo", "junio")

# ONI observado para febrero y marzo
oni_obs <- oni %>%
  ungroup() %>%
  filter(
    Year == year2,
    as.character(month) %in% c("febrero", "marzo")
  ) %>%
  mutate(
    month = factor(as.character(month), levels = meses_pred)
  ) %>%
  dplyr::select(Year, month, oni) %>%
  arrange(month)

# ONI pronosticado para abril, mayo y junio
oni_forecast <- tibble(
  Year = year2,
  month = factor(c("abril", "mayo", "junio"), levels = meses_pred),
  oni = round(as.numeric(oni.pron$mean), 1)
)

# Base completa para predecir febrero-junio
new.data <- bind_rows(oni_obs, oni_forecast) %>%
  mutate(
    month = factor(as.character(month), levels = meses_pred)
  ) %>%
  arrange(Year, month)

preds <- predict(
  conts.lm,
  newdata = new.data,
  interval = "confidence"
)

preds <- cbind(new.data, preds) %>%
  mutate(
    lwr = ceiling(lwr),
    fit = ceiling(fit),
    upr = ceiling(upr)
  ) %>%
  dplyr::select(month, lwr, fit, upr) %>%
  mutate(month = as.character(month))

preds <- bind_rows(
  preds,
  preds %>%
    summarise(
      month = "Total",
      lwr = sum(lwr, na.rm = TRUE),
      fit = sum(fit, na.rm = TRUE),
      upr = sum(upr, na.rm = TRUE)
    )
)

cat("\n============================\n")
cat("===== PREDICCIÓN DEL MODELO =====\n")
cat("============================\n")

preds

cat("\n ** Salio mismo resultado que modelo anterior ** \n")
```

### Propuesta EDVG con round para no inflar datos
#### Es el mismo modelo con predicción con round y no ceiling

```r
# Probar con round para no inflar las predicciones (EDVG)
pred_raw <- predict(conts.lm, newdata = new.data, interval = "confidence")

preds <- cbind(new.data["month"], as.data.frame(pred_raw)) %>%
  mutate(
    month = as.character(month),
    across(c(lwr, fit, upr), ~ pmax(0, round(.x)))
  )

preds <- dplyr::bind_rows(
  preds,
  preds %>%
    summarise(
      month = "Total",
      lwr = sum(lwr),
      fit = sum(fit),
      upr = sum(upr)
    )
  ) %>% dplyr::select("month", "lwr", "fit", "upr") %>% mutate(month = as.character(month))
```

```r
cat("\n Se uso round para no inflar datos \n")
cat("\n floor para limites inferiores \n")
cat("\n ceiling para limites superiores \n")

# Probar con round para no inflar las predicciones (EDVG)
# floor para limites inferiores hacia abajo
# ceiling limites superiores hacia arriba

pred_raw <- predict(
  conts.lm,
  newdata = new.data,
  interval = "confidence"
)

preds <- cbind(new.data["month"], as.data.frame(pred_raw)) %>%
  mutate(
    month = as.character(month),
    lwr = pmax(0, floor(lwr)),
    fit = pmax(0, round(fit)),
    upr = pmax(0, ceiling(upr))
  ) %>%
  dplyr::select(month, lwr, fit, upr)

preds <- bind_rows(
  preds,
  preds %>%
    summarise(
      month = "Total",
      lwr = sum(lwr, na.rm = TRUE),
      fit = sum(fit, na.rm = TRUE),
      upr = sum(upr, na.rm = TRUE)
    )
)

cat("\n============================\n")
cat("===== PREDICCIÓN DEL MODELO =====\n")
cat("============================\n")

preds
```

# Modelo con poisson para serie de tiempo
#### Con año + mes + oni

```r
mod_pois <- glm(
  ndias ~ Year + month + oni,
  data = oni.conts,
  family = poisson(link = "log") 
)

# Predice con escala logaritmica
```

```r
par(mfrow = c(2, 2))
plot(mod_pois)
par(mfrow = c(1, 1))

cat("\n ===== Residuals vs Fitted ===== \n")
cat("\n No esta capturando bien la estructura de los datos \n")
cat("\n puede ser porque tiene muchos ceros y pocos valores altos \n")

cat("\n ===== Q-Q Residuals ===== \n")
cat("\n Se desvia en la cola superior (inidca meses/años con residuos altos) \n")

cat("\n ===== Scale-Location ===== \n")
cat("\n La linea sube. Variabilidad de los residuos conforme aumentan los valores ajustados \n")

cat("\n ===== Residuals vs Leverage ===== \n")
cat("\n Hay puntos influyentes que estan jalando el modelo (149, 17, 16) \n")
```

```r
cat("\n===== Prueba de dispersión =====\n")

dispersion <- sum(residuals(mod_pois, type = "pearson")^2) / df.residual(mod_pois)
dispersion


cat("\n El modelo poisson no tiene sobredispersion \n")
cat("\n Buen modelo \n")
# El modelo poisson no tiene sobredispersion
# Buen modelo
```

### Evalución del modelo

```r
cat("\n===== Resumen del modelo =====\n")
summary(mod_pois)

cat("\n===== AIC =====\n")
AIC(mod_pois)

cat("\n===== BIC =====\n")
BIC(mod_pois)


cat("\n===== Pseudo R2 =====\n")
pseudo_R2 <- 1 - mod_pois$deviance / mod_pois$null.deviance
pseudo_R2

cat("\n===== Comparar vs modelo nulo =====\n")
anova(mod_pois, test = "Chisq")


cat("\n con Poisson ajusta mejor el modelo \n")

cat("\n Residual deviance baja mucho con respecto a Null deviance (bastante bueno) \n")
```

```r
# Predicción en escala link con error estándar
pred_link <- predict(
  mod_pois,
  newdata = new.data,
  type = "link",
  se.fit = TRUE
)

# Intervalo de confianza del 95%
z <- qnorm(0.975)


# Con exp(), regresa a la escala original

preds <- new.data %>%
  mutate(
    month = as.character(month),
    fit = exp(pred_link$fit),
    lwr = exp(pred_link$fit - z * pred_link$se.fit),
    upr = exp(pred_link$fit + z * pred_link$se.fit)
  ) %>%
  dplyr::select(month, lwr, fit, upr) %>%
  mutate(
    across(c(lwr, fit, upr), ~ round(.x))
  )

preds <- dplyr::bind_rows(
  preds,
  preds %>%
    summarise(
      month = "Total",
      lwr = sum(lwr),
      fit = sum(fit),
      upr = sum(upr)
    )
)

cat("\n=============================\n")
cat("----- PREDICCIÓN DEL MODELO -----\n")
cat("\n=============================\n")

preds

cat("\n en general capto mejor la estructura de los datos \n")
```

```r
# Carga de datos

year1 <- 2012
year2 <- as.numeric(format(Sys.Date(), "%Y")) #- 1
meses <- c("enero", "febrero", "marzo", "abril", "mayo", "junio", "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre")

# Carga de datos viento y temperatura 

temperatura <- evola.exp::redes_con(
  red = "redmet",
  parametros = "tmp",
  start = year1,
  end = year2
)




v_viento <- evola.exp::redes_con(
  red = "redmet",
  parametros = "wsp",
  start = year1,
  end = year2
)
# Promedio de viento

wsp.d <- v_viento %>%
  mutate(date = as.Date(date),
         Year = lubridate::year(date),
         mes = lubridate::month(date)) %>%
  group_by(Year, mes) %>%
  summarise(
    wsp = mean(valor, na.rm = T),
    wsp_max = max(valor, na.rm = T),
    .groups = "drop"
  ) %>% 
  dplyr::filter(Year != year2) %>% 
  mutate(month = month(mes, label = T, abbr = F))


# Variable de viento bajo

wsp_bajo <- v_viento %>%
  mutate(date = as.Date(date)) %>%
  group_by(date) %>%
  summarise(
    wsp_dia = mean(valor, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  mutate(
    Year = as.numeric(format(date, "%Y")),
    month = format(date, "%B")
  ) %>%
  group_by(Year, month) %>%
  summarise(
    ndias_wsp_bajo = sum(wsp_dia <= 1.5, na.rm = TRUE),
    wsp = mean(wsp_dia, na.rm = TRUE),
    .groups = "drop"
  )



# Media, maximo de temperaturas y conteo de dias de temperaturas

tmp.d <- temperatura %>%
  mutate(date = as.Date(date)) %>%
  
  # máximo diario
  group_by(date) %>%
  summarise(
    tmp_prom = mean(valor, na.rm = TRUE),
    tmp_max = max(valor, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  
  # agregación mensual
  group_by(
    Year = format(date, "%Y"),
    month = format(date, "%B")
  ) %>%
  
  summarise(
    
    ndias_28 = sum(tmp_max >= 28, na.rm = TRUE),
    ndias_29 = sum(tmp_max >= 29, na.rm = TRUE),
    ndias_30 = sum(tmp_max >= 30, na.rm = TRUE),
    ndias_31 = sum(tmp_max >= 31, na.rm = TRUE),
    ndias_32 = sum(tmp_max >= 32, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  
  filter(Year != year2)
```

```r
# Cargar datos ONI

oni <- read.csv("\\\\192.168.17.149\\Analisis\\ENSO\\Tablas\\enso.csv")



oni <- round(oni, 1) %>% 
  filter(YR >= 1990)

colnames(oni) <- c("Year", meses)
oni <- gather(oni, month, oni, -Year) %>% 
  dplyr::filter(Year >= year1) %>% 
  dplyr::arrange(Year)
```

```r
# Datos horarios O3
o3.d <- evola.exp::redes_con(start = year1, end = year2) %>% 
  evola.exp::filtrar_est(transporte = F) %>% 
  dplyr::group_by(date = as.Date(date)) %>% 
  dplyr::summarise(valor = max(valor, na.rm = T))
```

```r
## Conteo de dias >= 155 por año
o3.m <- o3.d %>%
  dplyr::group_by(Year = format(date, "%Y"), month = format(date, "%B")) %>%
  dplyr::summarise(ndias = sum(valor >= 155, na.rm = T)) %>% 
  dplyr::filter(Year != year2)
```

```r
## Datos para modelo de regresion

# Se agrega o3 con oni

oni.conts <- merge(oni, o3.m, sort = F) %>% 
  dplyr::mutate(month = factor(month, levels = meses))


# se agregan datos de tmp

oni.conts <- merge(oni.conts, tmp.d, sort = F) %>% 
  dplyr::mutate(month = factor(month, levels = meses))


# oni.conts <- merge(oni.conts, wsp.d, sort = F) %>% 
#   dplyr::mutate(month = factor(month, levels = meses))



# ONI por categorias

oni.conts <- oni.conts %>%
  mutate(
    oni_cat = case_when(
      oni >= 0.5 ~ "Nino",
      oni <= -0.5 ~ "Nina",
      TRUE ~ "Neutral"
    ),
    oni_cat = factor(oni_cat, levels = c("Nina", "Neutral", "Nino"))
  )


# Filtrar solo temporada de ozono
oni.temp <- oni.conts %>%
  filter(month %in% c("febrero","marzo","abril","mayo", "junio"))

# Agregar viento bajo

oni.temp <- oni.temp %>%
  left_join(wsp_bajo, by = c("Year", "month"))



# Modelo con wsp centrado

oni.temp <- oni.temp %>%
  mutate(
    wsp_c = wsp - mean(wsp, na.rm = TRUE)
  )

cat("\n Datos para el modelo \n")
head(oni.temp)
tail(oni.temp)

# Comparacion  de modelos con temperaturas
cat("\n Comparacion  de modelos con temperaturas \n")

mod_28 <- lm(ndias ~ Year + month + ndias_28 + wsp, data = oni.temp)
mod_29 <- lm(ndias ~ Year + month + ndias_29 + wsp, data = oni.temp)
mod_30 <- lm(ndias ~ Year + month + ndias_30 + wsp, data = oni.temp)
mod_31 <- lm(ndias ~ Year + month + ndias_31 + wsp, data = oni.temp) # mejor modelo
mod_32 <- lm(ndias ~ Year + month + ndias_32 + wsp, data = oni.temp)

comparacion <- data.frame(
  modelo = c("ndias_28", "ndias_29", "ndias_30", "ndias_31", "ndias_32"),
  RSE = c(summary(mod_28)$sigma,
          summary(mod_29)$sigma,
          summary(mod_30)$sigma,
          summary(mod_31)$sigma,
          summary(mod_32)$sigma),
  R2_ajustado = c(summary(mod_28)$adj.r.squared,
                  summary(mod_29)$adj.r.squared,
                  summary(mod_30)$adj.r.squared,
                  summary(mod_31)$adj.r.squared,
                  summary(mod_32)$adj.r.squared),
  AIC = c(AIC(mod_28),
          AIC(mod_29),
          AIC(mod_30),
          AIC(mod_31),
          AIC(mod_32)),
  BIC = c(BIC(mod_28),
          BIC(mod_29),
          BIC(mod_30),
          BIC(mod_31),
          BIC(mod_32))
)

comparacion

```

```r
# Revision de DF
head(oni.temp)
```

```r
# Modelo con interaccion
mod_inter_c <- lm(
  ndias ~ Year + month + ndias_31 * wsp_c,
  data = oni.temp
)

summary(mod_inter_c)
AIC(mod_31, mod_inter_c)
BIC(mod_31, mod_inter_c)
```

```r
# Probar GLM con poisoon


mod_pois <- glm(
  ndias ~ Year + month + ndias_31 * wsp_c,
  data = oni.temp,
  family = poisson()
)

summary(mod_pois)
```

```r
# Prueba binomial negativa



mod_nb <- glm.nb(
  ndias ~ Year + month + ndias_31 * wsp_c,
  data = oni.temp
)

summary(mod_nb)

# Comparación de modelo con poisson

AIC(mod_pois, mod_nb)
BIC(mod_pois, mod_nb)
```

```r
# Validación con datos temporales

train <- oni.temp %>%
  filter(Year <= 2018)

test <- oni.temp %>%
  filter(Year >= 2019)

wsp_mean_train <- mean(train$wsp, na.rm = TRUE)

train <- train %>%
  mutate(wsp_c = wsp - wsp_mean_train)

test <- test %>%
  mutate(wsp_c = wsp - wsp_mean_train)

```

```r
# Validacion temporal con regresion lineal binomial

mod_nb_train <- MASS::glm.nb(
  ndias ~ Year + month + ndias_31 + wsp_c,
  data = train
)

pred_nb <- predict(
  mod_nb_train,
  newdata = test,
  type = "response"
)

error <- test$ndias - pred_nb

data.frame(
  ME = mean(error, na.rm = TRUE),
  RMSE = sqrt(mean(error^2, na.rm = TRUE)),
  MAE = mean(abs(error), na.rm = TRUE)
)
```

```r
# Validacion temporal con regresion lineal normal
mod_train <- lm(
  ndias ~ Year + month + ndias_31 * wsp_c,
  data = train
)

pred <- predict(mod_train, newdata = test)

error <- test$ndias - pred

data.frame(
  ME = mean(error, na.rm = TRUE),
  RMSE = sqrt(mean(error^2, na.rm = TRUE)),
  MAE = mean(abs(error), na.rm = TRUE)
)
```

```r
year1 <- 2012

year2 <- as.numeric(format(Sys.Date(), "%Y")) 
meses <- c(
  "enero", "febrero", "marzo", "abril", "mayo", "junio",
  "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"
)

# Clasificación de meses para ONI
oni_meses <- c(
  "DJF" = "enero",
  "JFM" = "febrero",
  "FMA" = "marzo",
  "MAM" = "abril",
  "AMJ" = "mayo",
  "MJJ" = "junio",
  "JJA" = "julio",
  "JAS" = "agosto",
  "ASO" = "septiembre",
  "SON" = "octubre",
  "OND" = "noviembre",
  "NDJ" = "diciembre"
)



# Bases de ONI
oni <- read.table(
  "https://www.cpc.ncep.noaa.gov/data/indices/oni.ascii.txt",
  header = TRUE
) %>% 
  rename(Year = YR, meses = SEAS , oni = ANOM) %>% 
  filter(Year <= year2) %>% 
  dplyr::select(- TOTAL) %>% 
  dplyr::mutate(oni = round(oni, 1),
         month = oni_meses[meses],
         month = factor(month, levels = .env$meses))

oni <- oni %>%
  mutate(
    month_num = match(as.character(month), meses)
  ) %>%
  arrange(Year, month_num)

oni.ts <- ts(
  oni$oni,
  start = c(min(oni$Year), 1),
  frequency = 12
)

start(oni.ts)
end(oni.ts)

# Pronosticar ONI

oni.ts.mod <- auto.arima(oni.ts, seasonal = TRUE)

oni.pron <- forecast(oni.ts.mod, h = 3)

oni.pron

oni_mayo_2026 <- round(as.numeric(oni.pron$mean[2]), 1)
oni_mayo_2026
```
